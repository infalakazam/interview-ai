```python
import numpy as np
import pandas as pd
from datasets import load_dataset
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sentence_transformers import SentenceTransformer
import joblib
```

```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset(
    "nkazi/MohlerASAG",
    name="cleaned",
    split="open_ended"
)

df = dataset.to_pandas()

df.info()
```

```python
df.head()
```

```python
df["score_avg"].describe()
```

```python
df=df[["question",
    "instructor_answer",
    "student_answer",
    "score_avg"
]].copy()

```

```python
for col in ["question", "instructor_answer", "student_answer"]:
    df[col] = df[col].astype(str).str.strip()

df = df[
    (df["question"] != "") &
    (df["instructor_answer"] != "") &
    (df["student_answer"] != "")
].copy()

print("Dataset shape after cleaning:", df.shape)
```

```python
encoder = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
questions=df["question"].tolist()
reference_answers=df['instructor_answer'].tolist()
student_answers=df['student_answer'].tolist()
```

```python
question_embeddings=encoder.encode(questions,show_progress_bar=True,convert_to_numpy=True)
reference_embeddings=encoder.encode(reference_answers,show_progress_bar=True,convert_to_numpy=True)
student_embeddings=encoder.encode(student_answers,show_progress_bar=True,convert_to_numpy=True)
```

```python
print("Question embeddings:", question_embeddings.shape)
print("Reference embeddings:", reference_embeddings.shape)
print("Student embeddings:", student_embeddings.shape)
```

```python
X=np.concatenate([question_embeddings,reference_embeddings,student_embeddings],axis=1)
X.shape
```

```python
y=df["score_avg"].values
print("Target shape:", y.shape)
print("First 5 target scores:", y[:5])
```

```python
from sklearn.model_selection import GroupShuffleSplit
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
# Hold out complete questions so the test set measures generalization to unseen prompts.
groups = df["question"].astype(str).str.strip().str.lower()
train_idx, test_idx = next(splitter.split(X, y, groups=groups))
X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]
print("Train/test rows:", len(train_idx), len(test_idx))
print("Question overlap:", len(set(groups.iloc[train_idx]) & set(groups.iloc[test_idx])))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = np.sum(
    reference_embeddings * student_embeddings,
    axis=1
) / (
    np.linalg.norm(reference_embeddings, axis=1)
    * np.linalg.norm(student_embeddings, axis=1)
    + 1e-10
)

print("Similarity shape:", similarity.shape)
print("First 10 similarity values:", similarity[:10])
```

```python
embedding_difference = np.abs(
    reference_embeddings - student_embeddings
)

X_enhanced = np.concatenate(
    [
        question_embeddings,
        reference_embeddings,
        student_embeddings,
        embedding_difference,
        similarity.reshape(-1, 1)
    ],
    axis=1
)

print("Enhanced feature matrix:", X_enhanced.shape)
```

```python
X_train_enhanced = X_enhanced[train_idx]
X_test_enhanced = X_enhanced[test_idx]

print(X_train_enhanced.shape)
print(X_test_enhanced.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

enhanced_model = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

enhanced_model.fit(X_train_enhanced, y_train)

y_pred_enhanced = enhanced_model.predict(
    X_test_enhanced
)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae_enhanced = mean_absolute_error(
    y_test, y_pred_enhanced
)

mse_enhanced = mean_squared_error(
    y_test, y_pred_enhanced
)

rmse_enhanced = np.sqrt(mse_enhanced)

r2_enhanced = r2_score(
    y_test, y_pred_enhanced
)

print("Enhanced MAE:", mae_enhanced)
print("Enhanced MSE:", mse_enhanced)
print("Enhanced RMSE:", rmse_enhanced)
print("Enhanced R²:", r2_enhanced)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

ridge_model = make_pipeline(
    StandardScaler(),
    Ridge(alpha=10.0)
)

ridge_model.fit(X_train_enhanced, y_train)

y_pred_ridge = ridge_model.predict(X_test_enhanced)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae_ridge = mean_absolute_error(y_test, y_pred_ridge)
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mse_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)

print("Ridge MAE:", mae_ridge)
print("Ridge MSE:", mse_ridge)
print("Ridge RMSE:", rmse_ridge)
print("Ridge R²:", r2_ridge)
```

## Tuned scikit-learn model comparison

This section compares familiar regression models on the same enhanced embedding features. It tunes hyperparameters with `GroupKFold`, keeping all answers to a question together inside each validation fold. The held-out test questions are used only for the final comparison. Scores are clipped to the valid 0–5 range for reporting.

```python
from sklearn.model_selection import GroupKFold, RandomizedSearchCV
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd
import numpy as np

train_groups = groups.iloc[train_idx].to_numpy()
cv = GroupKFold(n_splits=min(5, len(np.unique(train_groups))))

candidates = {
    "Ridge": (
        make_pipeline(StandardScaler(), Ridge()),
        {"ridge__alpha": [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0]}
    ),
    "Random Forest": (
        RandomForestRegressor(random_state=42, n_jobs=1),
        {"n_estimators": [200, 400], "max_features": [0.5, 1.0],
         "min_samples_leaf": [1, 2, 4], "max_depth": [None, 20, 40]}
    ),
    "Extra Trees": (
        ExtraTreesRegressor(random_state=42, n_jobs=1),
        {"n_estimators": [200, 400], "max_features": [0.5, 1.0],
         "min_samples_leaf": [1, 2, 4], "max_depth": [None, 20, 40]}
    ),
    "Hist Gradient Boosting": (
        HistGradientBoostingRegressor(random_state=42),
        {"max_iter": [100, 200, 300], "learning_rate": [0.03, 0.06, 0.1],
         "max_leaf_nodes": [15, 31], "l2_regularization": [0.0, 0.1, 1.0]}
    ),
    "RBF SVR": (
        make_pipeline(StandardScaler(), SVR(kernel="rbf")),
        {"svr__C": [0.1, 1.0, 10.0, 50.0],
         "svr__epsilon": [0.05, 0.1, 0.25, 0.5],
         "svr__gamma": ["scale", 0.001, 0.01]}
    )
}

searches = {}
comparison_rows = []
for name, (estimator, param_distributions) in candidates.items():
    search = RandomizedSearchCV(
        estimator=estimator,
        param_distributions=param_distributions,
        n_iter=8,
        scoring="r2",
        cv=cv,
        random_state=42,
        n_jobs=-1,
        verbose=1,
        refit=True
    )
    search.fit(X_train_enhanced, y_train, groups=train_groups)
    searches[name] = search
    raw_prediction = search.best_estimator_.predict(X_test_enhanced)
    bounded_prediction = np.clip(raw_prediction, 0, 5)
    comparison_rows.append({
        "Model": name,
        "Best CV R2": search.best_score_,
        "Test R2 (bounded)": r2_score(y_test, bounded_prediction),
        "Test MAE": mean_absolute_error(y_test, bounded_prediction),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, bounded_prediction)),
        "Best parameters": search.best_params_
    })

comparison = pd.DataFrame(comparison_rows).sort_values("Test R2 (bounded)", ascending=False)
comparison[["Model", "Best CV R2", "Test R2 (bounded)", "Test MAE", "Test RMSE"]]
```

```python
# Choose hyperparameters using grouped CV, not the held-out test set.
best_name = comparison.sort_values("Best CV R2", ascending=False).iloc[0]["Model"]
best_model = searches[best_name].best_estimator_
print("Selected model by grouped CV:", best_name)
print("Best hyperparameters:", searches[best_name].best_params_)
print("Held-out R2 for selected model:", comparison.loc[comparison["Model"] == best_name, "Test R2 (bounded)"].iloc[0])
```

## Interpreting the results

R² is not guaranteed to improve just by switching models. The grouped split is stricter than a row-wise split because the model must score questions it did not see during training. Compare `Best CV R2` and `Test R2 (bounded)`: a large gap can indicate overfitting. Use MAE/RMSE as well, since they show the typical error in score points. For a product that will repeatedly score answers to known prompts, you can also run a separate row-wise split experiment, but label it clearly because it answers an easier question.

```python
# Fit the selected estimator on all cleaned Mohler rows for the local API.
# Test metrics above remain based only on held-out questions.
from sklearn.base import clone
import joblib

deployment_model = clone(best_model)
deployment_model.fit(X_enhanced, y)
artifact = {
    "model": deployment_model,
    "encoder_name": "all-MiniLM-L6-v2",
    "feature_version": 1,
    "target_min": 0.0,
    "target_max": 5.0,
    "dataset": "nkazi/MohlerASAG/open_ended",
    "training_rows": len(df),
}
joblib.dump(artifact, "prepai_mohler_model.joblib")
print("Saved prepai_mohler_model.joblib")
```
